````python
# Vraag 1a.1: de index bepalen.
# Uit printen van de z-array blijkt dat z = 0 niet expliciet ertussen staat, daarom hieronder code om de index van z te vinden waar
# de waarde het meest dichtbij 0 ligt

# Z gaat van negatief naar positief, we isoleren eerst de twee indexen waar de switch van positief naar negatief plaatsvindt 
# en bepalen dan welke dichter bij 0 ligt

for i in range(len(z)):
    if np.sign(z[i]) == np.sign(z[i+1]):
        continue
    val1 = i
    val2 = i + 1
    break

if np.abs(0-val1) < np.abs(0-val2):
    print("de index waarbij z het dichtst bij 0 ligt is", val1)
else:
    print("de index waarbij z het dichtst bij 0 ligt is", val2)
````

````python
#Vraag 1a.2: log van het potentiaal tegen de log van de straal

#De potentiaal is een 2D array en de straal 1D, we slicen de potentiaal array op z = 0 om ze tegen elkaar te kunnen plotten
Vplot = V[50, :]

# De eerste waarde van Vplot = 0, en daarvan kan geen logaritme genomen worden. We slicen de eerste waarde dus uit allebei arrays
Vplot = Vplot[1:]
rplot = r[1:]

plt.figure()
plt.plot(np.log(rplot), np.log(Vplot), '.')
plt.xlabel("log van radius (m)")
plt.ylabel("log van potentiaal op z = 0 (V)")
plt.show()
````

````python
# Vraag 1a.3: fit
from scipy.optimize import curve_fit

#verband lijkt lineair, dus daarop fitten
def fit(x, a, b):
    return a*x + b

vals, covs = curve_fit(fit, np.log(rplot), np.log(Vplot))
xtest = np.linspace(np.min(np.log(rplot)), np.max(np.log(rplot)), 500)
ytest = fit(xtest, *vals)

plt.figure()
plt.plot(xtest, ytest, label = 'fit')
plt.plot(np.log(rplot), np.log(Vplot), '.', label = 'data')
plt.xlabel("log van radius (m)")
plt.ylabel("log van potentiaal op z = 0 (V)")
plt.legend()
plt.show()
````

````python
#Vraag 1a.4: de exponent

#Bij de gebruikte fit is de exponent van r gelijk aan de helling van de lijn
print(f"De exponent is ({vals[0]:.3f} +/- {np.sqrt(covs[0,0]):.4g})")
````

#### Vraag 3c: De oppervlaktelading van een geleidende schijf op constant potentiaal

De oppervlakte lading $\sigma(r)$ van een geleidende schijf met straal $R$ is niet uniform verdeeld over de schijf, maar wordt gegeven door (Problems 2.58 and 2.63 Griffiths 5th edition):
$$\sigma(r) = \frac{Q}{2 \pi R \sqrt{R^2 - r^2}}$$

1. Bereken de totale lading op de schijf met behulp van vraag 3a en 3b.
2. Vergelijk de oppervlakte lading in het centrum van de geleidende schijf met dat van een uniform geladen schijf met dezelfde totale lading.
3. Plot de oppervlakte lading van de geleidende schijf met de berekende lading $Q$, tussen $r = 0$ en $R$.
4. Plot in hetzelfde figuur de oppervlaktelading van een uniform geladen schijf met dezelfde lading $Q$.


````python
# De totale lading op de schijf (indien de oppervlaktelading constant is) wordt gegeven door de oppervlaktelading * oppervlakte schijf

# Deelvraag 1
Qtot_sigma_constant = sigma_center * np.pi * disk_radius**2
print(f"De totale lading op de schijf als de oppervlakte lading constant is = {Qtot_sigma_constant:.4g} C")

# Oppervlaktelading in het centrum van de geleidende schijf
sigma_r = Qtot_sigma_constant / (2 * np.pi * disk_radius * np.sqrt(disk_radius**2))
print(f"Voor de uniform geleidende schijf is de oppervlaktelading in het centrum {sigma_center:.4g} C/m^2 en voor de geleidende schijf is het {sigma_r:.4g} C/m^2")

def opplading(Q, R, r):
    return Q / (2*np.pi*R*np.sqrt(R**2 - r**2))

# We zorgen ervoor dat de array niet volledig tot R gaat, want anders zal er een 0 in de noemer zitten wanneer de functie aangeroepen wordt
Rplot = np.linspace(0, disk_radius*0.999, 500)
sigma_plot = opplading(Qtot_sigma_constant, disk_radius, Rplot)

plt.figure()
plt.title("Oppervlaktelading tegen r")
plt.plot(Rplot, sigma_plot, "--", label="Oppervlaktelading van geleidende schijf")
plt.axhline(sigma_center, 0, disk_radius, label = "Oppervlaktelading van uniform geladen schijf")
plt.legend()
plt.xlabel("r (m)")
plt.ylabel("Oppervlaktelading (C/m^2)")
plt.show()
````
