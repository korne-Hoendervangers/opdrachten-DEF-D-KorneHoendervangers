```python
# Bereken hier je benodigde kantelfrequentie
#De Eigenfrquentie kan je vinden door de omega & omega_0 gelijk te stellen aan 1
Frequentie = 5 #hz
Frequentie_0 = Frequentie
print(Frequentie_0)
```

```python
# Voeg hier je plot in.
Frequentie = np.linspace(0,100,10000)
Amplitude = np.abs(1/(1-((2*np.pi*Frequentie)**2)/(Frequentie_0*2*np.pi)**2))

plt.figure()
plt.loglog(Frequentie,Amplitude,"k.")
plt.xlabel("Frequentie")
plt.ylabel('Amplitudeverhouding')
plt.title("Verwachte grafiek bij minimale Delta_X waarde")
plt.show()
```

```python
# Bereken hier de minimale Delta-x waarbij je aan de eis van de opdracht voldoet.

#A_m/A_v = 1/1-(omega)²/Omega_0² ,omschrijven geeft
K = factor = 5
Frequentie = 5
Omega_0_Kwadraat = (Frequentie*2*np.pi)**2/(K+1)
g = 9.81
Delta_x = g/Omega_0_Kwadraat
print(Delta_x,"Minimaal aantal meter indrukking")
```

## Opdracht 3: Plot de grafiek.

```python
# Voeg hier je plot in. Lees goed de opdracht wat er in de grafiek moet staan!
# Het is ook handig om je grafiek op te slaan, doe dat door met plt.savefig().

Massa = np.linspace(0,10,1000)
C_1 = (Massa*g)/Delta_x
C_2 = (Massa*g)/(Delta_x*2)
C_3 =(Massa*g)/(Delta_x*5)

plt.figure()
plt.plot(Massa,C_1,"k.")
plt.plot(Massa,C_2,"b.")
plt.plot(Massa,C_3,"r.")
plt.ylabel("Veerconstante (C)")
plt.title("Verhouding Massa & Veerconstante bij verschillende Delta_X")
plt.xlabel("Massa (kg)")
plt.savefig("plotDeltaX")
```

Iteraties: We willen dat rond de 1.5 hertz ook gedempt wordt.

- Extra veer op de balk zetten in plaats van op het platteau:
Ook dit gaf weinig verbetering op het interval waar wij willen.

- Kijken wat de veer op zichzelf doet:
De veer geeft hetzelfde resultaat als de elastieken, dus deze is ook niet interessant.

- Zowel de veer met touw, als het elastiek vastmaken aan de massa en kijken of degene die niks doet of versterkt weg valt: Bij trillingen komt het heel snel tot stilstand, maar daarbij hebben allebei wel veel grote trillingen.

- Elastiek ophangen aan enkel punt: Werkte niet goed, doordat de elastiek aan het einde van zijn elastische rek zat.

Problemen: 
- De elastiek rekt langzaam uit waardoor de metingen steeds onbetrouwbaarder worden.

- De massa beweegt ook heen en weer wanneer er lang gemeten wordt. dit zorgt ook voor onzekerheid.
