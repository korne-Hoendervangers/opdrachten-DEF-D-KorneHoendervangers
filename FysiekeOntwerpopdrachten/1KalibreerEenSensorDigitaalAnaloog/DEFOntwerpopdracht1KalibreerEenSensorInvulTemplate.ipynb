```python
# Zet hier jouw data in een voor jou handig formaat
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
arduino_waarde = np.array([121,265,280,480, 525,574, 522, 450, 450, 380,340,276,280,290,257,195,202,137])
gewicht = np.array([79.63,148.4,230.67,366.83,457.6,522.75,409.23,389.4,374.46,337.6,321.0,285.2,258.0,221,192.7,178,156,124.9])

onzekerheid_meter= 20  #bits
onzekerheid_platform = 50 #bits
onzekerheid = np.sqrt(onzekerheid_meter**2 + onzekerheid_platform**2)
```

```python
# Plot hier de meting en fit een lijn door de metingen.
# Maak gebruik van wat je geleerd hebt bij het practicum.
plt.figure()
plt.plot(gewicht,arduino_waarde,'o')
def line(x,a,b):
    return a*x+b
poptval,pcovval = curve_fit(line, gewicht, arduino_waarde)
x = np.linspace(min(gewicht),max(gewicht),100)
plt.plot(x,line(x,*poptval))
plt.errorbar(gewicht, arduino_waarde, yerr=onzekerheid, fmt='o')    
plt.xlabel('gewicht')
plt.ylabel('arduino waarde')
plt.title('gewicht vs arduino waarde')
plt.show()  
print('a:',poptval[0])
print('b:',poptval[1])
```
